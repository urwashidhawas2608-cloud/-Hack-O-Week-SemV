```python
# Simple function
def greet(name):
    return f"Hello, {name}!"

print(greet("Urwashi"))


# Function with multiple parameters
def calculate_average(marks):
    return sum(marks) / len(marks)

marks = [85, 90, 78, 92, 88]

average = calculate_average(marks)

print("Marks:", marks)
print("Average:", average)
```

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    else:
        return "D"


marks = [85, 90, 78, 92, 88]

print("Average:", calculate_average(marks))
print("Grade:", calculate_grade(marks))
```

```python
class Student:

    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Marks:", self.marks)
        print("Average:", self.average())


student1 = Student(
    "Urwashi",
    101,
    [85, 90, 78, 92, 88]
)

student1.display()
```

```python
data = {
    "Name": ["Aarav", "Riya", "Urwashi", "Kabir", "Ananya"],
    "Marks": [85, 92, np.nan, 76, 95],
    "Department": ["CS", "IT", "CS", "IT", "CS"]
}

df_clean = pd.DataFrame(data)

print("Original Data:")
print(df_clean)

print("\nMissing Values:")
print(df_clean.isnull().sum())
```

```python
# Fill missing marks with average marks
df_clean["Marks"] = df_clean["Marks"].fillna(
    df_clean["Marks"].mean()
)

print("Cleaned Data:")
print(df_clean)
```

```python
import pandas as pd
import numpy as np

student_data = {
    "Name": [
        "Aarav", "Riya", "Urwashi", "Kabir",
        "Ananya", "Rahul", "Sneha", "Aditya",
        "Meera", "Arjun"
    ],

    "Department": [
        "CS", "IT", "CS", "IT", "CS",
        "IT", "CS", "IT", "CS", "IT"
    ],

    "Study_Hours": [
        5, 7, 6, 3, 8,
        4, 7, 5, 9, 6
    ],

    "Attendance": [
        90, 95, 88, 70, 96,
        75, 92, 85, 98, 89
    ],

    "Marks": [
        78, 92, 85, 65, 95,
        70, 88, 80, 96, 84
    ]
}

students_df = pd.DataFrame(student_data)

students_df
```

```python
print("Dataset Shape:", students_df.shape)

print("\nAverage Marks:")
print(students_df["Marks"].mean())

print("\nHighest Marks:")
print(students_df["Marks"].max())

print("\nLowest Marks:")
print(students_df["Marks"].min())

print("\nAverage Marks by Department:")
print(
    students_df.groupby("Department")["Marks"].mean()
)
```

```python
marks = np.array(students_df["Marks"])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
```

```python
def performance_category(mark):
    if mark >= 90:
        return "Excellent"
    elif mark >= 80:
        return "Good"
    elif mark >= 70:
        return "Average"
    else:
        return "Needs Improvement"


students_df["Performance"] = [
    performance_category(mark)
    for mark in students_df["Marks"]
]

students_df
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=students_df,
    x="Study_Hours",
    y="Marks",
    hue="Department",
    s=100
)

plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=students_df,
    x="Performance"
)

plt.title("Student Performance Distribution")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

numeric_data = students_df[
    ["Study_Hours", "Attendance", "Marks"]
]

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Student Features")

plt.show()
```
