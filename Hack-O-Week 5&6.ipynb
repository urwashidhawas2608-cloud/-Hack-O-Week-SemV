```python
# Creating vectors
v1 = np.array([2, 3])
v2 = np.array([4, 1])

print("Vector 1:", v1)
print("Vector 2:", v2)

# Vector addition
addition = v1 + v2

# Vector subtraction
subtraction = v1 - v2

print("\nVector Addition:", addition)
print("Vector Subtraction:", subtraction)
```

```python
v = np.array([3, 4])

# Magnitude of vector
magnitude = np.linalg.norm(v)

# Scalar multiplication
scaled_vector = 2 * v

print("Vector:", v)
print("Magnitude:", magnitude)
print("Scaled Vector:", scaled_vector)
```

```python
matrix_A = np.array([
    [1, 2],
    [3, 4]
])

matrix_B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(matrix_A)

print("\nMatrix B:")
print(matrix_B)
```

```python
# Matrix addition
addition = matrix_A + matrix_B

# Matrix subtraction
subtraction = matrix_A - matrix_B

# Matrix multiplication
multiplication = np.matmul(matrix_A, matrix_B)

print("Matrix Addition:")
print(addition)

print("\nMatrix Subtraction:")
print(subtraction)

print("\nMatrix Multiplication:")
print(multiplication)
```

```python
# Student features
features = np.array([5, 90, 8])

# Model weights
weights = np.array([2, 0.5, 3])

prediction = np.dot(features, weights)

print("Features:", features)
print("Weights:", weights)
print("Prediction:", prediction)
```

```python
matrix = np.array([
    [4, 1],
    [2, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("Matrix:")
print(matrix)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
def f(x):
    return x ** 2

def derivative(x):
    return 2 * x

x = 5

print("f(x):", f(x))
print("Derivative at x =", x, ":", derivative(x))
```

```python
def f(x):
    return x ** 2

x = 5
h = 0.0001

numerical_derivative = (
    f(x + h) - f(x)
) / h

print("Approximate derivative:", numerical_derivative)
```

```python
def gradient(x, y):
    return np.array([
        2 * x,
        2 * y
    ])

x = 3
y = 4

result = gradient(x, y)

print("Point:", (x, y))
print("Gradient:", result)
```

```python
x = 8

learning_rate = 0.1

for i in range(10):

    gradient = 2 * x

    x = x - learning_rate * gradient

    print(
        "Step:", i + 1,
        "x =", round(x, 4)
    )
```

```python
def derivative_chain_rule(x):
    return 4 * (2 * x + 1)

x = 3

print("x =", x)
print("Derivative =", derivative_chain_rule(x))
```

```python
x = 2
w = 3

# Forward pass
z = w * x
y = z ** 2

# Backward pass
dy_dz = 2 * z
dz_dw = x

dy_dw = dy_dz * dz_dw

print("Input x:", x)
print("Weight w:", w)
print("z:", z)
print("Output y:", y)

print("\nGradient dy/dw:", dy_dw)
```

```python
# Function: f(x) = x^2
def function(x):
    return x ** 2

# Derivative: f'(x) = 2x
def derivative(x):
    return 2 * x

x = 10
learning_rate = 0.1

history = []

for i in range(20):

    gradient = derivative(x)

    history.append(function(x))

    x = x - learning_rate * gradient

print("Minimum approximately reached at x =", x)
print("Function value =", function(x))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history, marker="o")

plt.title("Gradient Descent Progress")
plt.xlabel("Iteration")
plt.ylabel("Function Value")

plt.grid()
plt.show()
```
